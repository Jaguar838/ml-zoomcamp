## Питання 2: Ймовірність конверсії

Напишемо скрипт для завантаження моделі та розрахунку ймовірності конверсії для клієнта.

```python
import sklearn
sklearn.__version__
```

```python
import pickle

# Шлях до файлу з моделлю
model_file = 'pipeline_v1.bin'

# 1. Завантаження моделі (конвеєра) з файлу
# 'rb' означає читання у бінарному режимі (read binary)
try:
    with open(model_file, 'rb') as f_in:
        pipeline = pickle.load(f_in)
    print(f"Модель '{model_file}' успішно завантажено.")
except FileNotFoundError:
    print(f"Помилка: Файл моделі '{model_file}' не знайдено.")
    print("Будь ласка, переконайтеся, що ви завантажили його за допомогою wget і він знаходиться в тій самій папці, що і скрипт.")
    exit()


# 2. Дані клієнта для оцінки
# Це ті самі дані, що вказані у домашньому завданні
client_record = {
    "lead_source": "paid_ads",
    "number_of_courses_viewed": 2,
    "annual_income": 79276.0
}

# 3. Отримання прогнозу
# Метод predict_proba повертає ймовірності для кожного класу [негативний, позитивний]
# Модель очікує на вхід список, тому ми передаємо [client_record]
prediction_probabilities = pipeline.predict_proba([client_record])[0]

# Ймовірність конверсії (позитивний клас, індекс 1)
conversion_probability = prediction_probabilities[1]

print(f"\nДані клієнта: {client_record}")
print(f"Ймовірність конверсії для цього клієнта: {conversion_probability:.3f}")

```

## Питання 4: Ймовірність конверсії (через веб-сервіс)

Тепер давайте запитаємо наш локальний веб-сервіс, щоб отримати прогноз для іншого клієнта.

```python
import requests

# URL вашого FastAPI сервісу
# Якщо ви запускаєте його локально, це буде http://localhost:8000
url = "http://localhost:8000/predict"

# Дані клієнта для оцінки, як зазначено в завданні
client = {
    "lead_source": "organic_search",
    "number_of_courses_viewed": 4,
    "annual_income": 80304.0
}

# Надсилаємо POST-запит до сервісу
response = requests.post(url, json=client).json()

# Виводимо отриману ймовірність
print(f"Ймовірність конверсії для цього клієнта: {response['conversion_probability']:.3f}")
```
